## 6. Primeras operaciones técnicas

En esta sección se implementará una función en Python para calcular un indicador descriptivo de la matrícula de los Institutos Profesionales de Chile durante 2025.

Se utilizará la biblioteca NumPy para calcular el promedio de matrícula total por registro de programa académico.

Esta operación permitirá comprobar el funcionamiento de las funciones, el procesamiento de datos y la obtención de indicadores estadísticos a partir de información real del CNED.


```python

# Función para calcular el promedio de matrícula por registro

def calcular_promedio_matricula(tabla):
    """Calcula el promedio de matrícula total por registro."""

    if tabla.empty:
        raise ValueError("La tabla no puede estar vacía")

    matricula = pd.to_numeric(
        tabla["Matrícula Total"],
        errors="coerce"
    )

    if matricula.notna().sum() == 0:
        raise ValueError("No existen valores válidos de matrícula")

    return float(np.nanmean(matricula))

# Ejecutar la función con los datos de IP de 2025
promedio_matricula = calcular_promedio_matricula(datos_ip)

print(f"Promedio de matrícula por registro: {promedio_matricula:.2f}")
```


## 7. Verificación inicial

En esta sección se realizarán pruebas para verificar el correcto funcionamiento de la función de cálculo de matrícula y la consistencia básica de los datos utilizados.

Se comprobará que:

1. El conjunto de datos contenga al menos 2.000 registros.
2. Todos los registros correspondan a Institutos Profesionales del año 2025.
3. La matrícula total no presente valores negativos.
4. La función de cálculo del promedio entregue un resultado consistente.
5. La función detecte correctamente un conjunto de datos vacío.


```python

# Verificar cantidad de registros
assert len(datos_ip) >= 2000, "Cantidad insuficiente de registros"

# Verificar tipo de institución y año
assert (datos_ip["Tipo Institución"] == "I.P.").all()
assert (datos_ip["Año"] == 2025).all()

# Verificar que la matrícula no tenga valores negativos
matricula = pd.to_numeric(
    datos_ip["Matrícula Total"],
    errors="coerce"
)

assert (matricula.dropna() >= 0).all(), \
    "Existen valores negativos de matrícula"

# Verificar el cálculo del promedio
promedio_esperado = matricula.mean()

assert np.isclose(promedio_matricula, promedio_esperado), \
    "El promedio calculado no coincide"

# Verificar que la función detecte tablas vacías
try:
    calcular_promedio_matricula(datos_ip.iloc[0:0])
    raise AssertionError("No se detectó la tabla vacía")
except ValueError:
    print("Prueba de tabla vacía superada")

print("Todas las verificaciones iniciales fueron superadas")
```
