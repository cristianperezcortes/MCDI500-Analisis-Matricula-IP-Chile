## 1. Introducción

En esta fase se realizará la exploración, limpieza, transformación y validación de los datos correspondientes a los Institutos Profesionales de Chile durante 2025.

Se utilizará Python y las bibliotecas Pandas y NumPy para identificar valores faltantes, revisar posibles inconsistencias, preparar las variables y generar un conjunto de datos procesado.

El análisis se desarrollará a partir de los 4.731 registros y 62 columnas identificados durante la Fase 1.


```python

# 2. Carga de la base de datos del CNED

from pathlib import Path
import pandas as pd
import numpy as np

# Ubicación del archivo Excel
archivo = Path("../data/raw/Base-INDICES-Matricula-2005-2026.xlsx")

# Comprobar que el archivo existe
print("¿Existe el archivo?:", archivo.exists())

# Cargar los datos
datos = pd.read_excel(archivo)

# Mostrar dimensiones de la base
print("Filas y columnas:", datos.shape)

# Mostrar los primeros cinco registros
display(datos.head())
```

```python

# 3. Filtrar Institutos Profesionales del año 2025

datos_ip = datos[
    (datos["Tipo Institución"] == "I.P.") &
    (datos["Año"] == 2025)
].copy()

# Mostrar resultados del filtro
print("Registros de IP en 2025:", len(datos_ip))

print("Cantidad de columnas:", datos_ip.shape[1])

print(
    "Cantidad de Institutos Profesionales:",
    datos_ip["Nombre Institución"].nunique()
)

# Mostrar los primeros cinco registros
display(datos_ip.head())
```

```python

# 4. Análisis de valores faltantes

# Contar valores faltantes por columna
faltantes = datos_ip.isnull().sum()

# Calcular porcentaje de valores faltantes
porcentaje = (faltantes / len(datos_ip)) * 100

# Crear tabla resumen
tabla_faltantes = pd.DataFrame({
    "Columna": datos_ip.columns,
    "Valores faltantes": faltantes.values,
    "Porcentaje (%)": porcentaje.values
})

# Mostrar solo columnas con valores faltantes
tabla_faltantes = tabla_faltantes[
    tabla_faltantes["Valores faltantes"] > 0
]

# Ordenar desde el mayor porcentaje
tabla_faltantes = tabla_faltantes.sort_values(
    by="Porcentaje (%)",
    ascending=False
)

print("Columnas con valores faltantes:", len(tabla_faltantes))

display(tabla_faltantes.round(2))
```

```python

# 5. Identificar columnas completamente vacías

columnas_vacias = datos_ip.columns[
    datos_ip.isna().all()
].tolist()

print("Cantidad de columnas completamente vacías:",
      len(columnas_vacias))

print("\nColumnas identificadas:")

for columna in columnas_vacias:
    print("-", columna)

# Calcular cuántas columnas quedarían
print("\nColumnas originales:", datos_ip.shape[1])
print("Columnas restantes si se excluyen:",
      datos_ip.shape[1] - len(columnas_vacias))
```

```python

# 6. Excluir columnas completamente vacías

# Crear una copia de los datos originales filtrados
datos_ip_limpios = datos_ip.copy()

# Eliminar columnas que no contienen información
datos_ip_limpios = datos_ip_limpios.drop(
    columns=columnas_vacias
)

# Comprobar los resultados
print("Columnas originales:", datos_ip.shape[1])
print("Columnas eliminadas:", len(columnas_vacias))
print("Columnas restantes:", datos_ip_limpios.shape[1])
print("Registros conservados:", len(datos_ip_limpios))

# Mostrar primeros registros
display(datos_ip_limpios.head())
```

```python

# 7. Revisar valores faltantes después de la limpieza inicial

faltantes_restantes = datos_ip_limpios.isna().sum()

porcentaje_faltantes = (
    faltantes_restantes / len(datos_ip_limpios)
) * 100

resumen_faltantes = pd.DataFrame({
    "Columna": datos_ip_limpios.columns,
    "Valores faltantes": faltantes_restantes.values,
    "Porcentaje (%)": porcentaje_faltantes.values
})

# Mostrar solamente columnas con valores faltantes
resumen_faltantes = resumen_faltantes[
    resumen_faltantes["Valores faltantes"] > 0
]

# Ordenar de mayor a menor
resumen_faltantes = resumen_faltantes.sort_values(
    "Porcentaje (%)",
    ascending=False
)

print("Columnas con valores faltantes:",
      len(resumen_faltantes))

display(resumen_faltantes.round(2))
```

```python

# 8. Revisar registros sin información de título

registros_sin_titulo = datos_ip_limpios[
    datos_ip_limpios["Título"].isna()
]

columnas_revision = [
    "Nombre Institución",
    "Nombre Programa",
    "Título",
    "Matrícula Total"
]

print("Cantidad de registros sin título:",
      len(registros_sin_titulo))

display(registros_sin_titulo[columnas_revision])
```

```python

# 9. Revisar variables con más del 90% de datos faltantes

columnas_revision = [
    "Grado Académico",
    "Mención o Especialidad"
]

for columna in columnas_revision:
    print("\nVariable:", columna)


    print("Registros con información:",
          datos_ip_limpios[columna].notna().sum())

    print("\nValores más frecuentes:")

    print(
        datos_ip_limpios[columna]
        .value_counts()
        .head(10)
    )

```

```python

# 11. Limpieza y transformación inicial de datos

# Crear copia para conservar los datos originales
datos_procesados = datos_ip_limpios.copy()

# Normalizar textos que representan valores faltantes
datos_procesados = datos_procesados.replace(
    ["Null", "NULL", "null", ""],
    np.nan
)

# Eliminar columnas con más del 90% de valores faltantes
porcentaje_nulos = datos_procesados.isna().mean() * 100

columnas_excluir = porcentaje_nulos[
    porcentaje_nulos > 90
].index.tolist()

datos_procesados = datos_procesados.drop(
    columns=columnas_excluir
)

# Eliminar espacios sobrantes en columnas de texto
columnas_texto = datos_procesados.select_dtypes(
    include=["object", "str"]
).columns

for columna in columnas_texto:
    datos_procesados[columna] = (
        datos_procesados[columna]
        .astype("string")
        .str.strip()
    )

# Verificar resultados
print("Registros:", len(datos_procesados))
print("Columnas restantes:", datos_procesados.shape[1])

print("\nColumnas excluidas:")
for columna in columnas_excluir:
    print("-", columna)

print("\nValores faltantes restantes:",
      datos_procesados.isna().sum().sum())

display(datos_procesados.head())
```

```python

# 12. Tratamiento de valores faltantes

# Crear una copia de seguridad antes de imputar
datos_antes_imputacion = datos_procesados.copy()

# Identificar columnas numéricas y de texto
columnas_numericas = datos_procesados.select_dtypes(
    include=["number"]
).columns

columnas_categoricas = datos_procesados.select_dtypes(
    include=["object", "string"]
).columns

# Completar datos numéricos faltantes con la mediana
# Solo en columnas que tienen valores disponibles
for columna in columnas_numericas:
    if datos_procesados[columna].isna().any():
        mediana = datos_procesados[columna].median()

        if pd.notna(mediana):
            datos_procesados[columna] = (
                datos_procesados[columna].fillna(mediana)
            )

# Identificar información categórica no informada
for columna in columnas_categoricas:
    datos_procesados[columna] = (
        datos_procesados[columna].fillna("No informado")
    )

# Verificar resultados
print("Registros:", len(datos_procesados))
print("Columnas:", datos_procesados.shape[1])

print("Valores faltantes antes:",
      datos_antes_imputacion.isna().sum().sum())

print("Valores faltantes después:",
      datos_procesados.isna().sum().sum())

display(datos_procesados.head())
```

```python

# 13. Transformación y validación de los datos

# Crear copia de los datos procesados
datos_finales = datos_procesados.copy()

# Reiniciar índices
datos_finales = datos_finales.reset_index(drop=True)

# Crear una variable categórica para el tamaño de matrícula
datos_finales["Categoría Matrícula"] = pd.cut(
    datos_finales["Matrícula Total"],
    bins=[-1, 50, 200, 1000, float("inf")],
    labels=["Baja", "Media", "Alta", "Muy alta"]
)

# Validaciones básicas
assert len(datos_finales) == 4731, "Cantidad incorrecta de registros"

assert datos_finales.isna().sum().sum() == 0, \
    "Existen valores faltantes"

assert (datos_finales["Matrícula Total"] >= 0).all(), \
    "Existen valores negativos de matrícula"

assert (datos_finales["Año"] == 2025).all(), \
    "Existen registros de otros años"

assert (datos_finales["Tipo Institución"] == "I.P.").all(), \
    "Existen instituciones que no son IP"

print("Validaciones completadas correctamente")
print("Registros:", len(datos_finales))
print("Columnas:", datos_finales.shape[1])
print("Valores faltantes:", datos_finales.isna().sum().sum())

print("\nDistribución por categoría de matrícula:")
print(datos_finales["Categoría Matrícula"].value_counts())

display(datos_finales.head())
```

```python

# 14. Exportar los datos procesados

from pathlib import Path

# Definir carpeta de destino
carpeta_destino = Path("../data/processed")

# Crear carpeta si no existe
carpeta_destino.mkdir(parents=True, exist_ok=True)

# Definir nombre del archivo
archivo_salida = carpeta_destino / "matricula_ip_2025_procesada.csv"

# Guardar archivo CSV
datos_finales.to_csv(
    archivo_salida,
    index=False,
    encoding="utf-8-sig"
)

# Comprobar que el archivo se guardó
print("Archivo guardado correctamente")
print("Ubicación:", archivo_salida)
print("Registros:", len(datos_finales))
print("Columnas:", datos_finales.shape[1])
print("¿Existe el archivo?:", archivo_salida.exists())
```

Esta imputación permite completar el procesamiento, pero los valores estimados no deben interpretarse como datos originales del CNED. Antes de obtener conclusiones sobre aranceles, vacantes u otros indicadores afectados, se deberá revisar su pertinencia.
